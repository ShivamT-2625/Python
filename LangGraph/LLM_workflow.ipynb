```python
from langgraph.graph import StateGraph,START,END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)
```

```python
#make the state
class LLMState(TypedDict):
    question:str
    answer:str
```

```python
def LLMResponse(state:LLMState)->LLMState:
    question = state["question"]
    #get a prompt about the process you wanna have
    prompt= f"Answer the following question {question}"

    #ask that question to the llm
    result = model.invoke(prompt).content
    #update the answer in the state
    state["answer"] = result
    return state
```

```python
#create the graph
graph = StateGraph(LLMState)

#add nodes to the graph
graph.add_node("LLMResponse",LLMResponse)

#add edges to the graph
graph.add_edge(START,"LLMResponse")
graph.add_edge("LLMResponse",END)

#compile the graph
workflow =graph.compile()
```

```python
initial_state = {"question":"Name of the continents in the world"}
final_state = workflow.invoke(initial_state)
print(final_state)
```
