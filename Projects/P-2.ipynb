## Finding out various things related to the data that i have in here


```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv("./used_cars.csv")
df.sample(5)
```

```python
df["brand"].unique()   
df["model_year"].unique()   
df["clean_title"].unique()  #nominal encoding 
```

```python
current_year = 2026
df["car_age"]=current_year - df["model_year"]
```

```python
df.sample(4)
df.drop(columns=["model_year"])
df['transmission'].unique()
```

```python
def clean_transmission (x):
   
    x = str(x).lower()

    if 'manual' in x or 'm/t' in x or 'mt' in x:
        return 'Manual'

    elif 'cvt' in x:
        return 'CVT'

    elif 'dual shift' in x or 'auto-shift' in x:
        return 'DualClutch'

    elif 'automatic' in x or 'a/t' in x:
        return 'Automatic'

    else:
        return 'Other'
```

```python
df["clean_transmission"] = df['transmission'].apply(clean_transmission)
df.sample(5)
df.drop(columns=["transmission"])
df["engine"].unique()
```

```python
#extract feature from data
df["horsepower"] = df["engine"].str.extract(r'(\d+\.?\d*)HP')[0]
#convert it to numeric
df['horsepower'] = pd.to_numeric(df["horsepower"])
df.sample(2)
df["horsepower"].isnull().sum()
```

```python
#extracting engine size
df["engine_size"]=df["engine"].str.extract(r'(\d+\.?\d*)L')[0]
df["engine_size"] = pd.to_numeric(df["engine_size"])
df.sample(2)
```

```python
df['cylinders_num'] = (
    df['cylinders']
    .str.extract(r'(\d+)')[0]
)
df["cylinders_num"] = pd.to_numeric(df["cylinders_num"])
```

```python
df.sample(2)
df.drop(columns=["cylinders"])
```

```python
def fueltype(x):
    x = str(x).lower()

    if 'diesel' in x:
        return 'Diesel'

    elif 'electric' in x:
        return 'Electric'

    elif 'hybrid' in x:
        return 'Hybrid'

    elif 'flex fuel' in x:
        return 'FlexFuel'

    else:
        return 'Gasoline'
```

```python
df["fuel_type"]= df["engine"].apply(fueltype)
df.sample(4)
df.drop(columns=["engine","transmission"],inplace=True)
```
