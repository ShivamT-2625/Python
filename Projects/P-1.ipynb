# Weather Dataset 

```python
import pandas as pd             
import matplotlib.pyplot as plt


df = pd.read_csv( "Sample - Superstore.csv" , encoding="latin1" )

print(df)
```

### Which product categories generates most sales vs profit?

```python
df.groupby("Category")[["Sales","Profit"]].sum().sort_values("Sales",ascending=False)

grouped = df.groupby("Category")[["Sales","Profit"]].sum()
grouped.plot(kind="bar")
plt.title("Sales vs Profit by Category")
plt.ylabel("Amount")
plt.show()

```

## Which States are underperforming (high sales but low profit or losses)?

```python
df.groupby("State")[["Sales","Profit"]].sum().sort_values("Profit",ascending=False)
```

```python
grouped = df.groupby("State")[["Sales","Profit"]].sum()
bottom_10 = grouped.sort_values("Profit",ascending=False).head(10)

bottom_10.plot(kind="bar")
plt.title("Sales vs Profit by State")
plt.ylabel("Amount")
plt.show()
```
