```python
import pandas as pd
import numpy as np


```

```python
import sklearn as sk

df = pd.read_excel("ML.xlsx")
print(df)
```

```python
from sklearn import linear_model
import matplotlib.pyplot as plt

plt.xlabel("area")
plt.ylabel("price")
plt.scatter(df.area,df.price)
```

```python
new_df = df.drop('price',axis="columns")
print(new_df)
```

```python
model = linear_model.LinearRegression()
print(model)
```

```python
model.fit(new_df, df.price)
```

```python
model.predict([[5000]])
```

```python
print(model.coef_)
model.intercept_
```

###Here the model.coef_ means the slope of the graph

#### The intercept is the y intercept_

y=mx+b

```python
area_df = pd.read_excel("Book1.xlsx")


p=model.predict(area_df)
area_df["prices"]=p

print(area_df)

area_df.to_csv("prediction.csv")
```

```python
read_df = pd.read_csv("hiring.csv")
print(read_df)
exp_map = {
    "zero": 0,
    "one": 1,
    "two": 2,
    "three": 3,
    "four": 4,
    "five": 5,
    "six": 6,
    "seven": 7,
    "eight": 8,
    "nine": 9,
    "ten": 10
}

read_df["experience"] = read_df["experience"].map(exp_map)
```

```python
read_df["experience"]=read_df["experience"].fillna(read_df["experience"].median())

read_df["test_score(out of 10)"]=read_df["test_score(out of 10)"].fillna(read_df["test_score(out of 10)"].median())
print(read_df)
```

```python
print(read_df)
```

```python
model2 = linear_model.LinearRegression()
read_df = read_df.rename(columns={"salary($)": "salary"})
```

```python
model2.fit(read_df.drop('salary',axis="columns"),read_df.salary)
```

```python
model2.predict([[2,9,6]])
model2.predict([[3,10,7]])
```

```python
import pickle
with open('model_pickle','wb') as file:
    pickle.dump(model2,file)
```
