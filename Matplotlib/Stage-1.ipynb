# Starting with Matplotlib




```python
import matplotlib.pyplot as plt

x = [ 1 ,2 ,3 ,4 ,5]
y= [2,1,1.75,2.6,3.5]

plt.plot(x,y)# x side for horizontal and y for vertical
# plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np 
```

```python
x =np.linspace(0,5,11)
x
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y,x)
```

```python
fig =plt.figure(figsize=(5,12),dpi=500)
fig =plt.figure()
axis1 = fig.add_axes([0.2,0.2,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.3,0.5,0.5])
axis2.plot(x,y)
```

```python
plt.hist(x)
plt.boxplot(x)
plt.scatter(x,y)
```

```python
fig = plt.figure(figsize=(5,12),dpi=500)
ax =fig.add_axes([0.2,0.2,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
