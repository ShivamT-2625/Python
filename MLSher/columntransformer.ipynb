## Without using column transformer

```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv("./covid_toy.csv")
df.sample(5)
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df["city"].value_counts()
```

```python
#Step1 obviously
from sklearn.model_selection import train_test_split
X_train ,X_test , y_train ,y_test = train_test_split(df.drop(columns=["has_covid"]),df["has_covid"],test_size=0.2,random_state=0)
```

# Usual Way 


```python
#adding simple imputer to fever col
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[["fever"]])
X_test_fever = si.transform(X_test[["fever"]])

```

```python
#ordinalEncoding -> cough
oe = OrdinalEncoder(categories=[["Mild","Strong"]])
X_train_cough = oe.fit_transform(X_train[["cough"]])
X_test_cough = oe.fit_transform(X_test[["cough"]])
X_train_cough
```

```python
#OneHotEncoding -> gender,city
ohe = OneHotEncoder(drop="first")
X_train_gender_city=ohe.fit_transform(X_train[["gender","city"]]).toarray()
X_test_gender_city = ohe.fit_transform(X_test[["gender","city"]]).toarray()
X_train_gender_city
```

```python
X_train_gender_city.shape
```

```python
#Extract Age
X_train_age = X_train.drop(columns=["gender","fever","cough","city"]).values
X_test_age = X_test.drop(columns=["gender","fever","cough","city"]).values
```

```python
X_train_age
```

```python
X_train_transformed = np.concatenate((X_train_age,X_train_fever,X_train_gender_city,X_train_cough),axis=1)
X_test_transformed = np.concatenate((X_test_age,X_test_fever,X_test_gender_city,X_test_cough),axis=1)
```

```python
X_train_transformed
```

## Faster way out 

```python
from sklearn.compose import ColumnTransformer
transformer =ColumnTransformer(transformers=[("tnf1",SimpleImputer(),["fever"]),("tnf2",OrdinalEncoder(categories=[["Mild","Strong"]]),["cough"]),("tnf3",OneHotEncoder(drop="first",sparse_output=False),["gender","city"])],remainder ="passthrough")#list mentioned here accepts tuple
```

```python
X_train_transformed_CT =transformer.fit_transform(X_train).shape
X_test_transformed_CT =transformer.fit_transform(X_test).shape
```
